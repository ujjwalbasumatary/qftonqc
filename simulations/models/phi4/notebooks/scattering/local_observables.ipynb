```julia
using Plots, JLD2, LaTeXStrings
```

# Energy density plots

```julia
data_lam_0 = load("./data/energy_scattering_infinite_dp_0p1_sigma_0p1_T_1000_mu0sq_0p5_lam_0p0_d_5_D_10_dt_0p1.jld2")
data_lam_2 = load("./data/energy_scattering_infinite_dp_0p1_sigma_0p1_T_1000_mu0sq_0p5_lam_2p0_d_5_D_10_dt_0p1.jld2")
```

```julia
dims = size(data_lam_0["energy_exp"])
dims2 = size(data_lam_2["energy_exp"])
```

```julia
p1 = heatmap(1:dims[2], 0.1 * [1:dims[1]], data_lam_0["energy_exp"], xlabel=L"Lattice site $n$", clim=(0,0.13), ylabel=L"Lattice time $t$", dpi = 600)
```

```julia
p2 = heatmap(1:dims2[2], 0.1 * [1:dims2[1]], data_lam_2["energy_exp"], clim=(0, 0.13), xlabel=L"Lattice site $n$", ylabel=L"Lattice time $t$", dpi = 600)
```

```julia
savefig(p1, "./energy_scattering_infinite_dp_0p1_sigma_0p1_T_1000_mu0sq_0p5_lam_0p0_d_5_D_10_dt_0p1.png")
```

```julia
savefig(p2, "./energy_scattering_infinite_dp_0p1_sigma_0p1_T_1000_mu0sq_0p5_lam_2p0_d_5_D_10_dt_0p1.png")
```
